## Експеримент
Повторяем опит. Старт 100, 36 месеца. Червеното е +30%. Не е прогноза.


```python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(7)
def months(kind, n=36, paths=80):
    if kind == 'normal':
        r = 0.007 + 0.045 * rng.normal(size=(paths, n))
    elif kind == 'student':
        r = 0.007 + 0.035 * rng.standard_t(5, size=(paths, n))
    else:
        r = 0.02 * rng.exponential(1.0, size=(paths, n))
    return np.clip(100 * np.cumprod(1 + r, axis=1), 5, None)
fig, axes = plt.subplots(1, 3, figsize=(12, 3.6), sharey=True)
for ax, kind, title in zip(axes, ('normal','student','exp'), ('нормално','t опашки','exp')):
    P = months(kind); t = np.arange(1, 37)
    ax.plot(t, P.T, color='#c8c0b2', lw=0.7)
    ax.plot(t, P[-1], color='#2d4a3e', lw=1.6)
    ax.axhline(130, color='#7a3b2e', ls='--', lw=1)
    ax.set_title(title); ax.set_xlabel('месец')
axes[0].set_ylabel('капитал')
fig.tight_layout(); plt.show()
print('t е по-близо до пазар. exp е само плюс.')
```
